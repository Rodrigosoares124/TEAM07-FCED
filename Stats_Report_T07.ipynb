## Statistical Analysis Before and After the 2017 Wildfire Crisis

This study examines wildfire ignition records from the 
Sistema de Gestão de Informação de Incêndios Florestais (SGIF), provided by the Instituto da Conservação da Natureza e das Florestas (ICNF).

It focuses on Braga, Guimarães, Guarda, and Faro between 2001 and 2025. The four municipalities represent different geographic, demographic, and territorial contexts in Portugal.

The analysis asks whether the distribution of human-caused ignitions and the annual number of fire ignitions changed after the 2017 wildfire crisis. It distinguishes between negligent and intentional ignitions and uses both descriptive and inferential statistical methods.

## Research Questions

### Main Research Question

Did the distribution of human-caused fire ignitions and the annual number of fire ignitions differ between the periods before and after 2017 in Braga, Guimarães, Guarda, and Faro?

### Specific research questions

#### RQ1: Proportion of Intentional Ignitions

Did the proportion of intentional ignitions among human-caused ignitions change significantly between the pre-2017 period (2001-2016) and the post-2017 period (2018-2025)?

**Response variable:** Cause type  
**Grouping variable:** Period  
**Statistical test:** Two-proportion test

#### RQ2: Annual Number of Ignitions

Did the mean annual number of fire ignitions differ significantly between the pre-2017 and post-2017 periods?

**Response variable:** Annual number of ignitions  
**Grouping variable:** Period  
**Statistical test:** Independent-samples t-test

#### RQ3: Differences Between Municipalities

Did the mean annual number of fire ignitions differ significantly among Braga, Guimarães, Guarda, and Faro?

**Response variable:** Annual number of ignitions  
**Grouping variable:** Municipality  
**Statistical test:** One-way ANOVA

All hypothesis tests use a significance level of (alpha = 0.05). The year 2017 is treated as a transition year and excluded from the before-and-after comparisons.

## Data and Study Scope

The dataset contains individual wildfire ignition records from the SGIF/ICNF database. The available files cover the period from 2001 to 2025.

The files are read and combined into a single dataset for the analysis of Braga, Guimarães, Guarda, and Faro. Municipality names are normalized so that differences in spelling and accents do not affect the selection.

**Source and provenance.** The data are the ICNF/SGIF *Registos individuais de incêndios 2001 a 2025* (Parts 1, 2 and 3), covering 2001-2025.

### Imports and Analysis Parameters

Column names are converted to uppercase, which ensures consistent processing across files from different years.

```python
from pathlib import Path
import pandas as pd

DATA_FOLDER = Path("dados")
MUNICIPALITIES = ["BRAGA", "GUIMARAES", "GUARDA", "FARO"]

MUNICIPALITIES_AREA_KM2 = {
    "BRAGA": 183,
    "GUIMARAES": 241,
    "GUARDA": 712,
    "FARO": 203,
}

```

### Loading and Combining the Data

The files are located in the `dados` folder. Each file is read separately, its column names are standardized, and the resulting tables are concatenated into a single dataset.

The output below gives the number of records and the available variables before the study scope is applied.

```python

files = sorted(
    file
    for extension in ("csv", "xlsx")
    for file in DATA_FOLDER.glob(f"*.{extension}")
)

if not files:
    raise FileNotFoundError(f"No files found in folder '{DATA_FOLDER}'.")

tables = []

for file in files:
    print("Reading:", file)

    if file.suffix.lower() == ".csv":
        table = pd.read_csv(
            file,
            sep=";",
            decimal=",",
            encoding="utf-8-sig",
            low_memory=False
        )
    else:
        table = pd.read_excel(file)

    table.columns = table.columns.astype(str).str.upper().str.strip()
    tables.append(table)

df = pd.concat(tables, ignore_index=True)

print(f"Total records: {len(df)}")
print("Columns:", sorted(list(df.columns)))
```

### Defining the Study Population

Municipality names are converted to uppercase, whitespace is removed, and accented characters are normalized. The resulting identifier is used to select the four municipalities included in the study.

The dataset is then restricted to Braga, Guimarães, Guarda, and Faro.

```python
# Standardize municipality names by converting to uppercase, trimming spaces,
# and removing accents/special characters to create a consistent matching key.
df["MUNICIPALITY"] = (
    df["CONCELHO"].astype(str).str.upper().str.strip()
    .str.normalize("NFKD").str.encode("ascii", "ignore").str.decode("ascii")
)

# Identify the municipality whose normalized versions
# match the reference municipality scope and display each unique matches.
matching_spellings = df.loc[
    df["MUNICIPALITY"].isin(MUNICIPALITIES),
    "CONCELHO"
].unique()
print("Matching spellings:", matching_spellings)


# Filter records belonging to the municipalities scope
# and create an independent copy of the resulting subset.
scoped_data = df[df["MUNICIPALITY"].isin(MUNICIPALITIES)].copy()
print(f"Records in scope: {len(scoped_data)}")
```

### Defining the Analysis Year

For each ignition, the analysis year is taken from the alert date whenever possible. If the alert date is unavailable or cannot be used, the original `ANO` variable provides the year.

This common year variable is used to define the periods for the statistical comparisons.

```python
# Create a unified year variable from the alert date when available;
# otherwise, use the existing year column.
if "DATAHORAALERTA" in scoped_data.columns:
    scoped_data["ANALYSIS_YEAR"] = pd.to_datetime(
        scoped_data["DATAHORAALERTA"], errors="coerce", dayfirst=True
    ).dt.year
else:
    scoped_data["ANALYSIS_YEAR"] = scoped_data["ANO"]


# Preview the municipality and analysis year fields.
scoped_data[
    [
        "MUNICIPALITY",
        "ANALYSIS_YEAR"
    ]
].head() #.head() #.tail()
```

### Normalizing Ignition Causes

Cause values are converted to uppercase. In the SGIF data, an empty cause field means the fire was **not investigated**, while `DESCONHECIDA` means it was investigated but the cause could not be determined.

Since the research questions focus only on human-caused ignitions (negligent vs. intentional), both situations are grouped into a single category, `DESCONHECIDA`, and flagged with `MISSING_CAUSE`. These records are excluded from the cause comparison but kept in the dataset and reported, because removing them silently could bias the analysis.

```python
# Clean and standardize cause descriptions, flag missing values,
# and assign a default label to records without cause information.
cause = (
    scoped_data["TIPOCAUSA"].astype(str).str.strip()
    .replace({"": None, "nan": None, "None": None})
)
scoped_data["CAUSE"] = cause.fillna("DESCONHECIDA").str.upper()
scoped_data["MISSING_CAUSE"] = scoped_data["CAUSE"].eq("DESCONHECIDA")

# Preview the original, normalized, and missing-cause indicators.
scoped_data[
    [
        "TIPOCAUSA",
        "CAUSE",
        "MISSING_CAUSE"
    ]
].head()
```

### Creating Derived Variables

The analysis uses four derived variables:

- `INTERVENTION_TIME_MINUTES`: time between the alert and the first intervention;
- `PERIOD`: temporal classification of each record;
- `HOUR_OF_DAY`: hour of the ignition alert, from 0 (midnight) to 23;
- `TIME_OF_DAY_PERIOD`: six-hour period of the ignition alert, grouping the hour into MADRUGADA, MANHÃ, TARDE, or NOITE.

| Period | Years | Role in the analysis |
|---|---:|---|
| Pre-2017 | 2001-2016 | Before the wildfire crisis |
| 2017 | 2017 | Transition year, excluded from comparisons |
| Post-2017 | 2018-2025 | After the wildfire crisis |

The hour of day is extracted from `DATAHORAALERTA`; the original `HORA` field is used only as a fallback when the alert timestamp is unavailable or invalid. The hour is then grouped into four chronological six-hour periods: MADRUGADA (00:00-05:59), MANHÃ (06:00-11:59), TARDE (12:00-17:59), and NOITE (18:00-23:59).

Ignition density is calculated from the area of each municipality so that comparisons account for differences in municipality size.


```python
alert_time = pd.to_datetime(scoped_data["DATAHORAALERTA"], errors="coerce", dayfirst=True)
intervention_time = pd.to_datetime(
    scoped_data["DATAHORA_PRIMEIRAINTERVENCAO"], errors="coerce", dayfirst=True
)
scoped_data["INTERVENTION_TIME_MINUTES"] = (intervention_time - alert_time).dt.total_seconds() / 60

# Derive the hour of day from the alert timestamp. The original HORA field
# is only a fallback when the timestamp is missing or invalid.
hour_from_alert = alert_time.dt.hour.astype("Int64")
hour_from_hora = pd.to_numeric(scoped_data["HORA"], errors="coerce")
valid_hora = hour_from_hora.between(0, 23) & hour_from_hora.mod(1).eq(0)
hour_from_hora = hour_from_hora.where(valid_hora).astype("Int64")
scoped_data["HOUR_OF_DAY"] = hour_from_alert.fillna(hour_from_hora)

# Aggregate the hour into four ordered six-hour periods. This keeps enough
# observations per bar for the before-and-after comparison.
TIME_OF_DAY_PERIODS = ["MADRUGADA", "MANHÃ", "TARDE", "NOITE"]
TIME_OF_DAY_BINS = [-1, 5, 11, 17, 23]
scoped_data["TIME_OF_DAY_PERIOD"] = pd.cut(
    scoped_data["HOUR_OF_DAY"].astype("Float64"),
    bins=TIME_OF_DAY_BINS,
    labels=TIME_OF_DAY_PERIODS,
)

# Audit the extraction and the agreement with the original HORA field.
hour_comparable = hour_from_alert.notna() & hour_from_hora.notna()
hour_extraction_audit = pd.DataFrame(
    {
        "Measure": [
            "Records in scope",
            "Hour extracted from DATAHORAALERTA",
            "Hour recovered from HORA fallback",
            "Missing or invalid hour",
            "Disagreements between DATAHORAALERTA and HORA",
        ],
        "Records": [
            len(scoped_data),
            int(hour_from_alert.notna().sum()),
            int((hour_from_alert.isna() & hour_from_hora.notna()).sum()),
            int(
                (
                    scoped_data["HOUR_OF_DAY"].isna()
                    | ~scoped_data["HOUR_OF_DAY"].between(0, 23).fillna(False)
                ).sum()
            ),
            int(
                (
                    hour_from_alert[hour_comparable]
                    != hour_from_hora[hour_comparable]
                ).sum()
            ),
        ],
    }
)

scoped_data["PERIOD"] = pd.cut(
    scoped_data["ANALYSIS_YEAR"], bins=[2000, 2016, 2017, 2025],
    labels=["2001-2016", "2017", "2018-2025"]
)

scoped_data[
    [
        "ANALYSIS_YEAR",
        "HOUR_OF_DAY",
        "TIME_OF_DAY_PERIOD",
        "INTERVENTION_TIME_MINUTES",
        "PERIOD",
    ]
].head()
```

```python
# Calculate the ignition density for a given scope of data.
# This function takes a DataFrame of ignition data and calculates
# the density of ignitions per square kilometer per year for each municipality.
def calculate_ignition_density(
    data,
    start_year=None,
    end_year=None,
    municipalities=MUNICIPALITIES,
    area_by_municipality=MUNICIPALITIES_AREA_KM2,
    year_column="ANALYSIS_YEAR",
):
    scoped = data.copy()

    if start_year is not None:
        scoped = scoped[scoped[year_column] >= start_year]

    if end_year is not None:
        scoped = scoped[scoped[year_column] <= end_year]

    years_in_scope = scoped[year_column].dropna().nunique()

    if years_in_scope == 0:
        raise ValueError("The selected scope contains no valid years.")

    result = (
        scoped.groupby("MUNICIPALITY")
        .size()
        .rename("IGNITIONS")
        .reindex(municipalities, fill_value=0)
        .to_frame()
    )

    result["AREA_KM2"] = result.index.map(area_by_municipality)
    result["YEARS"] = years_in_scope
    result["IGNITION_DENSITY_IGNITIONS_KM2_YEAR"] = (
        result["IGNITIONS"]
        / result["AREA_KM2"]
        / result["YEARS"]
    )

    return result.reset_index()
```

# Exploring the Data

This exploratory analysis characterizes the wildfire ignition records before the inferential stage. Its purpose is to describe the main features of the data, identify potential differences between the two periods, and provide descriptive evidence to guide the investigation of the research questions.

Four qualitative variables are analyzed in this section:

1. **Municipality.** Identifies the municipality where the ignition occurred. The study covers Braga, Guimarães, Guarda, and Faro.
2. **Period.** Classifies each ignition as pre-2017, 2017 (transition year), or post-2017. It is the grouping variable for every before-and-after comparison, so it is described through the period comparisons of the other variables rather than in a separate subsection.
3. **Cause type.** Identifies the recorded cause of the ignition. The composition charts compare every recorded cause category across periods, with particular attention to intentional and negligent ignitions, the two categories relevant to RQ1.
4. **Time of day (six-hour period).** Six-hour period of the ignition alert: MADRUGADA (00:00-05:59), MANHÃ (06:00-11:59), TARDE (12:00-17:59), or NOITE (18:00-23:59). It is analyzed within the cause-type study to compare the distribution of ignition times between the two periods and to examine how cause composition varies across these periods.

Four quantitative variables are analyzed in the subsequent section:

1. **Total burned area (ha).** Total area burned by each ignition, measured in hectares.
2. **Intervention time (minutes).** Elapsed time between the alert and the first intervention.
3. **Duration (hours).** Total duration of the fire, measured in hours.
4. **Ignition density (ignitions/km²/year).** Annual number of ignitions relative to the municipality's geographical area.

The descriptive comparisons use two periods:

- **Pre-2017:** 2001-2016.
- **Post-2017:** 2018-2025.

The year 2017 is treated as a transition year and is excluded from all before-and-after comparisons. The qualitative analysis first examines the distribution of ignitions across municipalities, then compares the relative composition of recorded causes across periods, and finally examines how cause composition varies across the four time-of-day periods.

The code cell below defines the shared color and formatting conventions used throughout the exploratory analysis; the same conventions will be reused for the quantitative variables.


```python
# Shared settings for the exploratory analysis.

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.ticker import PercentFormatter

# Fixed period colors and labels. These are reused in every period comparison.
COMPARISON_PERIODS = ["2001-2016", "2018-2025"]
PERIOD_LABELS = {"2001-2016": "2001-2016", "2018-2025": "2018-2025"}
PERIOD_COLORS = {"2001-2016": "#1F4E79", "2018-2025": "#FF7F0E"}

# Fixed cause order, colors, and labels. All recorded cause categories are shown.
CAUSE_COMPARISON = [
    "INTENCIONAL",
    "NEGLIGENTE",
    "REACENDIMENTO",
    "NATURAL",
    "DESCONHECIDA",
]
CAUSE_LABELS = {
    "INTENCIONAL": "Intentional",
    "NEGLIGENTE": "Negligent",
    "REACENDIMENTO": "Reignition",
    "NATURAL": "Natural",
    "DESCONHECIDA": "Unknown",
}
CAUSE_COLORS = {
    "INTENCIONAL": "#D62728",
    "NEGLIGENTE": "#1F4E79",
    "REACENDIMENTO": "#2CA02C",
    "NATURAL": "#FF7F0E",
    "DESCONHECIDA": "#7F7F7F",
}
CAUSE_TEXT_COLORS = {"DESCONHECIDA": "#1A1A1A"}
CAUSE_LABEL_MIN_PERCENT = 2.5

# Fixed time-of-day display labels. The period order is defined when the
# variable is created, so every plot uses the same chronological sequence.
TIME_OF_DAY_LABELS = {
    "MADRUGADA": "Madrugada\n(00:00-05:59)",
    "MANHÃ": "Manhã\n(06:00-11:59)",
    "TARDE": "Tarde\n(12:00-17:59)",
    "NOITE": "Noite\n(18:00-23:59)",
}

# Municipality order, display labels, and colors.
MUNICIPALITY_LABELS = {
    "BRAGA": "Braga",
    "GUIMARAES": "Guimarães",
    "GUARDA": "Guarda",
    "FARO": "Faro",
}
MUNICIPALITY_COLORS = {
    "BRAGA": "#1B9E77",
    "GUIMARAES": "#7570B3",
    "GUARDA": "#D95F02",
    "FARO": "#E7298A",
}
MUNICIPALITY_MARKERS = {
    "BRAGA": "o",
    "GUIMARAES": "s",
    "GUARDA": "^",
    "FARO": "D",
}

# Consistent academic figure style.
plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "axes.labelsize": 11,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.25,
    "grid.linewidth": 0.8,
    "legend.frameon": False,
})


def format_categorical_description(series, variable_name):
    """Return a readable one-variable description with two-decimal numbers."""
    summary = series.describe(include="all").to_frame(name=variable_name)
    summary = summary.rename(index={
        "count": "Count",
        "unique": "Unique values",
        "top": "Most frequent category",
        "freq": "Frequency",
    })
    summary[variable_name] = summary[variable_name].map(
        lambda value: f"{value:.2f}" if pd.api.types.is_number(value) else value
    )
    summary.index.name = "Statistic"
    return summary


def format_numeric_description(series, variable_name):
    """Return describe(include='all') with readable labels and two-decimal values."""
    summary = series.describe(include="all").to_frame(name=variable_name)
    summary = summary.rename(index={
        "count": "Count",
        "mean": "Mean",
        "std": "Standard deviation",
        "min": "Minimum",
        "25%": "25th percentile",
        "50%": "Median (50%)",
        "75%": "75th percentile",
        "max": "Maximum",
    })
    summary.index.name = "Statistic"
    summary[variable_name] = summary[variable_name].map(
        lambda value: f"{value:.2f}" if pd.api.types.is_number(value) else value
    )
    return summary


def missingness_summary(series, variable_name):
    """Return the non-missing and missing sample sizes for one variable."""
    total = int(series.shape[0])
    missing = int(series.isna().sum())
    available = total - missing
    return pd.DataFrame(
        {
            "Non-missing observations": [available],
            "Missing observations": [missing],
            "Missing (%)": [f"{100 * missing / total:.2f}"],
        },
        index=pd.Index([variable_name], name="Variable"),
    )


# The exploratory plots require complete keys; fail explicitly if any are missing.
if scoped_data[["MUNICIPALITY", "PERIOD", "CAUSE"]].isna().any().any():
    raise ValueError(
        "The scoped dataset contains missing municipality, period, or cause values."
    )
```

### Total Ignitions by Municipality and Period

The grouped bars compare the total number of recorded ignitions in each municipality between the two periods. The full in-scope ignition population is used, not only records with a known cause.


```python
# Count all in-scope ignitions by municipality and comparison period.
municipality_period_counts = (
    scoped_data.loc[scoped_data["PERIOD"].isin(COMPARISON_PERIODS)]
    .groupby(["MUNICIPALITY", "PERIOD"], observed=True)
    .size()
    .rename("IGNITIONS")
    .reindex(
        pd.MultiIndex.from_product(
            [MUNICIPALITIES, COMPARISON_PERIODS],
            names=["MUNICIPALITY", "PERIOD"],
        ),
        fill_value=0,
    )
    .reset_index()
)

if set(municipality_period_counts["MUNICIPALITY"]) != set(MUNICIPALITIES):
    raise ValueError("The municipality summary does not contain all four municipalities.")
if set(municipality_period_counts["PERIOD"]) != set(COMPARISON_PERIODS):
    raise ValueError("The municipality summary contains periods outside the comparison.")

municipality_counts = (
    municipality_period_counts
    .pivot(index="MUNICIPALITY", columns="PERIOD", values="IGNITIONS")
    .reindex(index=MUNICIPALITIES, columns=COMPARISON_PERIODS)
)

fig, ax = plt.subplots(figsize=(10.5, 6.2), constrained_layout=True)
positions = list(range(len(MUNICIPALITIES)))
bar_width = 0.38
max_count = municipality_counts.to_numpy().max()

for i, period in enumerate(COMPARISON_PERIODS):
    offset = (i - 0.5) * bar_width
    bar_positions = [position + offset for position in positions]
    bars = ax.bar(
        bar_positions,
        municipality_counts[period],
        width=bar_width,
        color=PERIOD_COLORS[period],
        edgecolor="white",
        linewidth=0.7,
        label=PERIOD_LABELS[period],
    )
    for bar, value in zip(bars, municipality_counts[period]):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + max_count * 0.012,
            f"{int(value):,}",
            ha="center",
            va="bottom",
            fontsize=8.5,
        )

ax.set_xticks(positions)
ax.set_xticklabels([MUNICIPALITY_LABELS[m] for m in MUNICIPALITIES])
ax.set_xlabel("Municipality")
ax.set_ylabel("Number of ignitions")
ax.set_title("Total ignitions by municipality and period")
ax.set_ylim(0, max_count * 1.16)
ax.grid(axis="y", alpha=0.25)
ax.set_axisbelow(True)
ax.legend(title="Period", loc="upper right")
plt.show()
plt.close(fig)
```

**NOTE:** Counts include all in-scope ignition records in the four municipalities. The 2017 transition year is excluded from both periods, and the number above each bar is the number of ignitions represented by that bar.

This chart is shown mainly to make clear that the two periods are not directly comparable in totals. The pre-2017 period covers 16 years (2001–2016) and the post-2017 period only 8 (2018–2025), so the totals for 2001–2016 are inflated simply because they include twice as many years. For example, Guimarães records 7,635 ignitions before 2017 and 730 after, but this partly reflects the different period lengths, even though it also matches the decresing trend from the before to the after 2017 period.

## Cause Type

Cause type is the qualitative variable used in RQ1. The full cause distribution is described first, followed by two complementary visualizations. The stacked bar plot focuses exclusively on human-caused ignitions (intentional and negligent), allowing their relative proportions to be compared across municipalities and study periods. The heatmaps, on the other hand, retain all recorded cause categories, providing a broader overview of the full cause distribution, including unknown causes.

This focus on human-caused ignitions is justified by the fact that the research questions underlying this study specifically address human-caused ignitions. Therefore, the stacked bar plot includes only the two relevant categories: intentional and negligent fires. The remaining categories are excluded from this visualization because they do not directly address the research questions:

- *Reignition (REACENDIMENTO)* is a rekindling of a previous fire, not a new ignition;
- *Natural (NATURAL)* causes are not human-caused by definition.

However, in order to visualize how the data available is distributed across all causes, we also provide a second plot (Heatmap) to examine the full cause composition across municipalities and study periods, providing additional context for the analysis.

### Relative composition of Ignition Causes

The comparison uses all records in the canonical dataset whose period is either 2001-2016 or 2018-2025. The canonical dataset is not modified.

The two visualizations use different denominators. The stacked bar chart focuses on human-caused ignitions and retains only `INTENCIONAL` and `NEGLIGENTE`, the two categories of primary interest for RQ1. Within each municipality and period, the proportion for each of these causes is calculated as:

$$
\text{Human-caused proportion} = \frac{\text{Number of ignitions with that human cause}}{\text{Total number of human-caused ignitions in the municipality-period}}
$$

The denominator is therefore the number of intentional plus negligent ignitions in that municipality and period, and the two proportions sum to 100%.

The heatmaps below retain all five recorded cause categories (`INTENCIONAL`, `NEGLIGENTE`, `REACENDIMENTO`, `NATURAL`, and `DESCONHECIDA`) and use all recorded ignitions in the municipality-period as the denominator, so the five proportions sum to 100%. This keeps the full cause composition visible, including the share of fires with no investigated cause.


```python
# Derived dataset for the cause composition. The canonical scoped_data is not modified.
cause_comparison_data = scoped_data.loc[
    scoped_data["PERIOD"].isin(COMPARISON_PERIODS)
].copy()

if cause_comparison_data[["MUNICIPALITY", "PERIOD", "CAUSE"]].isna().any().any():
    raise ValueError("The cause-comparison subset contains missing key values.")
if (cause_comparison_data["PERIOD"].astype(str) == "2017").any():
    raise ValueError("The 2017 transition year must be excluded from the cause comparison.")
if set(cause_comparison_data["MUNICIPALITY"]) != set(MUNICIPALITIES):
    raise ValueError("The cause-comparison subset does not contain all four municipalities.")
unexpected_causes = set(cause_comparison_data["CAUSE"]) - set(CAUSE_COMPARISON)
if unexpected_causes:
    raise ValueError(f"Unexpected cause categories were found: {sorted(unexpected_causes)}")

# Complete municipality-period-cause grid, preserving the fixed ordering.
cause_counts = (
    cause_comparison_data
    .groupby(["MUNICIPALITY", "PERIOD", "CAUSE"], observed=True)
    .size()
    .rename("IGNITIONS")
    .reindex(
        pd.MultiIndex.from_product(
            [MUNICIPALITIES, COMPARISON_PERIODS, CAUSE_COMPARISON],
            names=["MUNICIPALITY", "PERIOD", "CAUSE"],
        ),
        fill_value=0,
    )
    .reset_index()
)
cause_counts["TOTAL"] = (
    cause_counts.groupby(["MUNICIPALITY", "PERIOD"])["IGNITIONS"].transform("sum")
)
if (cause_counts["TOTAL"] == 0).any():
    raise ValueError("At least one municipality-period has no recorded ignitions.")
cause_counts["PROPORTION"] = cause_counts["IGNITIONS"] / cause_counts["TOTAL"]

proportion_sums = cause_counts.groupby(["MUNICIPALITY", "PERIOD"])["PROPORTION"].sum()
if (proportion_sums - 1).abs().max() > 1e-9:
    raise ValueError("Cause proportions do not sum to 100% within each municipality-period.")

print(
    f"Cause-comparison subset: {len(cause_comparison_data):,} records "
    f"across {cause_comparison_data['MUNICIPALITY'].nunique()} municipalities "
    f"and {cause_comparison_data['PERIOD'].nunique()} periods, "
    f"with {len(CAUSE_COMPARISON)} cause categories."
)

# Human-caused subset used by the stacked bar chart. The all-cause table above
# is kept intact for the heatmaps, which retain every recorded category.
HUMAN_CAUSES = ["INTENCIONAL", "NEGLIGENTE"]

human_cause_counts = cause_counts.loc[
    cause_counts["CAUSE"].isin(HUMAN_CAUSES)
].copy()
human_cause_counts["HUMAN_TOTAL"] = (
    human_cause_counts
    .groupby(["MUNICIPALITY", "PERIOD"], observed=True)["IGNITIONS"]
    .transform("sum")
)
if (human_cause_counts["HUMAN_TOTAL"] == 0).any():
    raise ValueError(
        "At least one municipality-period has no human-caused ignitions."
    )
human_cause_counts["HUMAN_PROPORTION"] = (
    human_cause_counts["IGNITIONS"] / human_cause_counts["HUMAN_TOTAL"]
)

human_proportion_sums = (
    human_cause_counts
    .groupby(["MUNICIPALITY", "PERIOD"], observed=True)["HUMAN_PROPORTION"]
    .sum()
)
if (human_proportion_sums - 1).abs().max() > 1e-9:
    raise ValueError(
        "Human-caused proportions do not sum to 100% within each "
        "municipality-period."
    )

print(
    f"Human-caused subset: {int(human_cause_counts['IGNITIONS'].sum()):,} "
    f"records out of {int(cause_counts['IGNITIONS'].sum()):,} recorded in the "
    f"comparison periods, with {len(HUMAN_CAUSES)} cause categories."
)
```

**NOTE:** Percentages in the stacked bars use human-caused ignitions only — intentional plus negligent — as the denominator (n shown above each bar), so the two segments sum to 100% of the human-caused total. Records classified as `REACENDIMENTO`, `NATURAL`, or `DESCONHECIDA` are excluded from these bars; their shares of all recorded ignitions are shown in the heatmaps below. The 2017 transition year is excluded.

## Time of Day

Time of day is derived from the ignition alert timestamp `DATAHORAALERTA`. The original `HORA` field is used only as a fallback when the alert timestamp is unavailable or invalid. To avoid very small hourly groups, the 24 hours are aggregated into four chronological six-hour periods:

|---|---|
| MADRUGADA | 00:00-05:59 |
| MANHÃ | 06:00-11:59 |
| TARDE | 12:00-17:59 |
| NOITE | 18:00-23:59 |

The analysis compares the distribution of human-caused ignitions across these four periods between 2001-2016 and 2018-2025 and examines how the relative composition of intentional and negligent ignitions varies between them. The 2017 transition year is excluded from these comparisons.


#### Descriptive Statistics

The summary below uses all records in the two comparison periods with a valid hour and reports the number and share of all recorded ignitions in each of the four time-of-day periods. It provides the overall time-of-day context; the stacked bar plot below restricts the comparison to human-caused ignitions. Missing or invalid hours are reported separately and are not included in the visualizations.

#### Distribution of Human-caused Ignitions by Time of Day

The two panels compare the distribution of human-caused ignitions across the four six-hour time-of-day periods in 2001-2016 and 2018-2025. Only intentional and negligent ignitions are included, as these are the two human-caused categories relevant to the research questions. The aim is to explore how the relative composition of human-caused ignitions varies throughout the day.


```python
# Human-caused composition within each six-hour time-of-day period.
# The denominator is intentional + negligent ignitions in that time-of-day
# period and comparison period; all-cause totals are not used here.
if (time_of_day_comparison_data["PERIOD"].astype(str) == "2017").any():
    raise ValueError("2017 must be excluded from the time-of-day comparison.")

time_of_day_human_data = time_of_day_comparison_data.loc[
    time_of_day_comparison_data["CAUSE"].isin(HUMAN_CAUSES)
].copy()
if set(time_of_day_human_data["CAUSE"]) != set(HUMAN_CAUSES):
    raise ValueError(
        "The time-of-day subset must contain both human-caused categories."
    )

time_of_day_human_counts = (
    time_of_day_human_data
    .groupby(["PERIOD", "TIME_OF_DAY_PERIOD", "CAUSE"], observed=True)
    .size()
    .rename("IGNITIONS")
    .reindex(
        pd.MultiIndex.from_product(
            [COMPARISON_PERIODS, TIME_OF_DAY_PERIODS, HUMAN_CAUSES],
            names=["PERIOD", "TIME_OF_DAY_PERIOD", "CAUSE"],
        ),
        fill_value=0,
    )
    .reset_index()
)
time_of_day_human_counts["HUMAN_TOTAL"] = (
    time_of_day_human_counts
    .groupby(["PERIOD", "TIME_OF_DAY_PERIOD"], observed=True)["IGNITIONS"]
    .transform("sum")
)
if (time_of_day_human_counts["HUMAN_TOTAL"] == 0).any():
    raise ValueError(
        "At least one time-of-day period has no human-caused ignitions."
    )
time_of_day_human_counts["HUMAN_PROPORTION"] = (
    time_of_day_human_counts["IGNITIONS"]
    / time_of_day_human_counts["HUMAN_TOTAL"]
)

time_of_day_human_sums = (
    time_of_day_human_counts
    .groupby(["PERIOD", "TIME_OF_DAY_PERIOD"], observed=True)["HUMAN_PROPORTION"]
    .sum()
)
if ((time_of_day_human_sums - 1).abs() > 1e-9).any():
    raise ValueError(
        "At least one time-of-day bar does not sum to 100% of the "
        "human-caused ignitions."
    )

time_of_day_values = np.arange(len(TIME_OF_DAY_PERIODS))
TIME_OF_DAY_BAR_WIDTH = 0.55
fig, axes = plt.subplots(
    1, 2, figsize=(13, 5.8), sharey=True, constrained_layout=True
)

for ax, period in zip(axes, COMPARISON_PERIODS):
    panel = (
        time_of_day_human_counts.loc[
            time_of_day_human_counts["PERIOD"] == period
        ]
        .pivot(
            index="TIME_OF_DAY_PERIOD",
            columns="CAUSE",
            values="HUMAN_PROPORTION",
        )
        .reindex(index=TIME_OF_DAY_PERIODS, columns=HUMAN_CAUSES, fill_value=0)
    )
    period_totals = (
        time_of_day_human_counts.loc[
            time_of_day_human_counts["PERIOD"] == period
        ]
        .groupby("TIME_OF_DAY_PERIOD", observed=True)["HUMAN_TOTAL"]
        .first()
        .reindex(TIME_OF_DAY_PERIODS, fill_value=0)
    )
    bottom = np.zeros(len(TIME_OF_DAY_PERIODS))

    for cause in HUMAN_CAUSES:
        percentages = panel[cause].to_numpy() * 100
        ax.bar(
            time_of_day_values,
            percentages,
            bottom=bottom,
            width=TIME_OF_DAY_BAR_WIDTH,
            color=CAUSE_COLORS[cause],
            edgecolor="white",
            linewidth=0.8,
        )

        # Label every non-empty segment at its vertical centre.
        for x_position, segment_bottom, percentage in zip(
            time_of_day_values, bottom, percentages
        ):
            if percentage > 0:
                ax.text(
                    x_position,
                    segment_bottom + percentage / 2,
                    f"{percentage:.1f}%",
                    ha="center",
                    va="center",
                    color=CAUSE_TEXT_COLORS.get(cause, "white"),
                    fontsize=9.5,
                    fontweight="bold",
                )

        bottom += percentages

    if (np.abs(bottom - 100) > 1e-6).any():
        raise ValueError(
            f"Human-caused time-of-day bars do not sum to 100% for {period}."
        )

    period_n = int((time_of_day_human_data["PERIOD"] == period).sum())
    ax.set_title(
        f"{PERIOD_LABELS[period]} (n = {period_n:,})",
        fontsize=12,
    )
    ax.set_xticks(time_of_day_values)
    ax.set_xticklabels(
        [TIME_OF_DAY_LABELS[period_name] for period_name in TIME_OF_DAY_PERIODS],
        fontsize=9.5,
    )
    ax.set_xlim(-0.6, len(TIME_OF_DAY_PERIODS) - 0.4)
    ax.set_ylim(0, 112)
    ax.set_yticks([0, 25, 50, 75, 100])
    ax.set_xlabel("Time of day")
    ax.axhline(
        100,
        color="#555555",
        linewidth=0.8,
        linestyle=":",
    )
    ax.grid(axis="y", alpha=0.25)
    ax.set_axisbelow(True)

    for x_position, total in zip(time_of_day_values, period_totals):
        ax.text(
            x_position,
            102,
            f"n = {int(total):,}",
            ha="center",
            va="bottom",
            fontsize=8.5,
            color="#333333",
        )

axes[0].set_ylabel(
    "Percentage of human-caused ignitions within the time-of-day period (%)"
)

legend_handles = [
    Patch(
        facecolor=CAUSE_COLORS[cause],
        edgecolor="white",
        label=CAUSE_LABELS[cause],
    )
    for cause in HUMAN_CAUSES
]
fig.legend(
    handles=legend_handles,
    title="Cause",
    loc="outside lower center",
    ncol=2,
    frameon=False,
)
fig.suptitle(
    "Distribution of human-caused ignitions by time of day",
    fontsize=13,
    fontweight="bold",
)

plt.show()
plt.close(fig)

print(
    f"Time-of-day human-caused table: {len(time_of_day_human_counts):,} rows "
    f"({len(COMPARISON_PERIODS)} periods × {len(TIME_OF_DAY_PERIODS)} "
    f"time-of-day periods × {len(HUMAN_CAUSES)} causes). "
    "All bars sum to 100% of the human-caused ignitions."
)
```

Each stacked bar represents 100% of the human-caused ignitions recorded in that time-of-day period. The segments show the relative proportions of intentional and negligent ignitions, and the percentage labels allow direct comparison between the two causes across different times of day. The sample size above each bar is the combined number of intentional and negligent ignitions in that time-of-day period and comparison period.

## Total Burned Area (ha)

Total burned area (`AREATOTAL_HA`) is the area burned by each ignition record, measured in hectares. The descriptive statistics and the two visualizations below use the `2001-2016` and `2018-2025` comparison records; the 2017 transition year is excluded from these before-and-after comparisons.


```python
# Shared helpers and comparison subset for the quantitative exploratory analysis.

quantitative_comparison_data = scoped_data.loc[
    scoped_data["PERIOD"].isin(COMPARISON_PERIODS)
].copy()

if (quantitative_comparison_data["PERIOD"].astype(str) == "2017").any():
    raise ValueError("2017 must be excluded from the before-and-after comparisons.")
if set(quantitative_comparison_data["PERIOD"].astype(str)) != set(COMPARISON_PERIODS):
    raise ValueError("The comparison subset contains unexpected periods.")

print(
    f"Quantitative comparison subset: {len(quantitative_comparison_data):,} records "
    f"across {quantitative_comparison_data['MUNICIPALITY'].nunique()} municipalities "
    f"and {quantitative_comparison_data['PERIOD'].nunique()} periods (2017 excluded)."
)


def to_log1p(values):
    """Map non-negative values to log10(1 + x) for display only."""
    return np.log10(1 + np.asarray(values, dtype=float))


def set_log1p_ticks(ax, tick_values, axis="x"):
    """Place original-unit tick labels on a log10(1 + x) axis."""
    tick_values = np.asarray(tick_values, dtype=float)
    positions = np.log10(1 + tick_values)
    labels = [
        f"{value:,.0f}" if value >= 1000 else f"{value:g}"
        for value in tick_values
    ]
    if axis == "x":
        low, high = ax.get_xlim()
        visible = (positions >= low - 1e-9) & (positions <= high + 1e-9)
        ax.set_xticks(positions[visible])
        ax.set_xticklabels([label for label, keep in zip(labels, visible) if keep])
    elif axis == "y":
        low, high = ax.get_ylim()
        visible = (positions >= low - 1e-9) & (positions <= high + 1e-9)
        ax.set_yticks(positions[visible])
        ax.set_yticklabels([label for label, keep in zip(labels, visible) if keep])
    else:
        raise ValueError("axis must be 'x' or 'y'")
```

### Distribution by Period

The two histograms compare the distribution of burned area in `2001-2016` and `2018-2025`, pooling the four municipalities within each period.


```python
# Histograms of total burned area by period, using identical bins and axes.
burned_area_histogram_data = quantitative_comparison_data.loc[
    :, ["PERIOD", "AREATOTAL_HA"]
].dropna(subset=["AREATOTAL_HA"])

burned_area_transformed = to_log1p(burned_area_histogram_data["AREATOTAL_HA"])
BURNED_AREA_BINS = 30
burned_area_bin_edges = np.linspace(
    0, burned_area_transformed.max(), BURNED_AREA_BINS + 1
)

burned_area_hist_max = max(
    np.histogram(
        burned_area_transformed[
            burned_area_histogram_data["PERIOD"] == period
        ],
        bins=burned_area_bin_edges,
    )[0].max()
    for period in COMPARISON_PERIODS
)

# Fit a normal distribution in the plotted log1p space and scale its density
# by n * bin width so that it is comparable to the histogram frequencies.
BURNED_AREA_BIN_WIDTH = float(np.diff(burned_area_bin_edges)[0])
burned_area_normal_curves = {}
for period in COMPARISON_PERIODS:
    period_values = burned_area_transformed[
        burned_area_histogram_data["PERIOD"] == period
    ]
    normal_mean = float(period_values.mean())
    normal_std = float(period_values.std(ddof=1))
    normal_x = np.linspace(
        burned_area_bin_edges[0], burned_area_bin_edges[-1], 400
    )
    normal_y = (
        np.exp(-0.5 * ((normal_x - normal_mean) / normal_std) ** 2)
        / (normal_std * np.sqrt(2 * np.pi))
        * len(period_values)
        * BURNED_AREA_BIN_WIDTH
    )
    burned_area_normal_curves[period] = (normal_x, normal_y)

burned_area_plot_max = max(
    burned_area_hist_max,
    max(curve.max() for _, curve in burned_area_normal_curves.values()),
)

fig, axes = plt.subplots(
    1, 2, figsize=(12, 5.2), sharex=True, sharey=True, constrained_layout=True
)
for ax, period in zip(axes, COMPARISON_PERIODS):
    values = burned_area_transformed[
        burned_area_histogram_data["PERIOD"] == period
    ]
    ax.hist(
        values,
        bins=burned_area_bin_edges,
        color=PERIOD_COLORS[period],
        edgecolor="white",
        linewidth=0.6,
    )
    normal_x, normal_y = burned_area_normal_curves[period]
    ax.plot(
        normal_x,
        normal_y,
        color="red",
        linestyle="-",
        linewidth=1.6,
        label="Normal distribution",
    )
    ax.set_title(f"{PERIOD_LABELS[period]} (n = {len(values):,})")
    ax.set_xlabel("Total burned area (ha; log10(1 + x) scale)")
    ax.set_xlim(burned_area_bin_edges[0], burned_area_bin_edges[-1])
    ax.set_ylim(0, burned_area_plot_max * 1.12)
    ax.legend(frameon=False)
    ax.grid(axis="y", alpha=0.25)
    ax.set_axisbelow(True)
    set_log1p_ticks(ax, [0, 0.1, 1, 10, 100, 1000, 10000], axis="x")

axes[0].set_ylabel("Number of ignitions")
fig.suptitle(
    "Distribution of total burned area by period",
    fontsize=13,
    fontweight="bold",
)
# plt.savefig("figures/distrubution_of_ignitions.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)
```

**NOTE:** Both panels use 30 equal-width bins on the `log10(1 + burned area)` scale and share the same x- and y-axis limits. The logarithmic display is justified by the strong right skew of burned area (median 0.08 ha; maximum 7,802.31 ha); it is a visualization choice only, and the original hectares are retained. No observations or outliers were removed. The 1,956 records with zero burned area remain in the first bin.

## Ignition Density (ignitions/km²/year)

Ignition density is an aggregated measure at the municipality-year level: the annual number of ignitions divided by the municipality's geographical area. It is constructed in a separate municipality-year table and is not merged into the row-level canonical dataset. Unlike the previous comparisons, this analysis includes all years from 2001 through 2025, including the 2017 transition year.


### Derived Municipality-Year table

The table below aggregates annual ignition counts by municipality and year, attaches the existing municipality areas, and calculates annual density as 

$$
\text{Ignition density (ignitions/km}^2\text{/year)} = \frac{\text{number of ignitions in the municipality-year}}{\text{area of the municipality (km}^2\text{)}}
$$

It uses the full ignition population, not only intentional and negligent causes. The existing `calculate_ignition_density` helper is not used here because it divides by the number of years and therefore returns a multi-year average; the annual table below divides each municipality-year count by area only.


```python
# Derived municipality-year table for annual ignition density.
expected_years = list(range(2001, 2026))
observed_years = set(
    scoped_data["ANALYSIS_YEAR"].dropna().astype(int).unique()
)
if observed_years != set(expected_years):
    raise ValueError("The canonical data do not cover every year from 2001 to 2025.")

annual_ignition_counts = (
    scoped_data
    .groupby(["MUNICIPALITY", "ANALYSIS_YEAR"], observed=True)
    .size()
    .rename("IGNITIONS")
    .reset_index()
)

expected_index = pd.MultiIndex.from_product(
    [MUNICIPALITIES, expected_years],
    names=["MUNICIPALITY", "ANALYSIS_YEAR"],
)
annual_ignition_counts = (
    annual_ignition_counts
    .set_index(["MUNICIPALITY", "ANALYSIS_YEAR"])
    .reindex(expected_index)
)
if annual_ignition_counts["IGNITIONS"].isna().any():
    missing_units = annual_ignition_counts.index[
        annual_ignition_counts["IGNITIONS"].isna()
    ].tolist()
    raise ValueError(
        "Missing municipality-year units were found and were not imputed as zero: "
        f"{missing_units}"
    )
annual_ignition_counts = annual_ignition_counts.reset_index()

annual_ignition_counts["AREA_KM2"] = (
    annual_ignition_counts["MUNICIPALITY"].map(MUNICIPALITIES_AREA_KM2)
)
if annual_ignition_counts["AREA_KM2"].isna().any():
    raise ValueError("At least one municipality has no area in MUNICIPALITIES_AREA_KM2.")

annual_ignition_counts["IGNITION_DENSITY_IGNITIONS_KM2_YEAR"] = (
    annual_ignition_counts["IGNITIONS"] / annual_ignition_counts["AREA_KM2"]
)

if len(annual_ignition_counts) != len(MUNICIPALITIES) * len(expected_years):
    raise ValueError("The annual density table does not have one row per municipality-year.")
if (annual_ignition_counts["IGNITION_DENSITY_IGNITIONS_KM2_YEAR"] < 0).any():
    raise ValueError("Negative ignition density values were found.")

# Independent spot checks of the aggregation and area division.
density_spot_checks = [
    ("BRAGA", 2001),
    ("FARO", 2017),
    ("GUARDA", 2025),
    ("GUIMARAES", 2010),
]
for municipality, year in density_spot_checks:
    expected_density = (
        ((scoped_data["MUNICIPALITY"] == municipality) & (scoped_data["ANALYSIS_YEAR"] == year)).sum()
        / MUNICIPALITIES_AREA_KM2[municipality]
    )
    actual_density = float(
        annual_ignition_counts.loc[
            (annual_ignition_counts["MUNICIPALITY"] == municipality)
            & (annual_ignition_counts["ANALYSIS_YEAR"] == year),
            "IGNITION_DENSITY_IGNITIONS_KM2_YEAR",
        ].iloc[0]
    )
    if not np.isclose(actual_density, expected_density):
        raise ValueError(
            f"Density spot check failed for {municipality} in {year}: "
            f"{actual_density} != {expected_density}."
        )

print(
    f"Annual density table: {len(annual_ignition_counts):,} municipality-year rows "
    f"for {len(MUNICIPALITIES)} municipalities and {len(expected_years)} years."
)
print("Independent density spot checks passed for: " + ", ".join(
    f"{municipality} {year}" for municipality, year in density_spot_checks
) + ".")
display(annual_ignition_counts.head(8).round(2))
```

## General Conclusion

- **Fewer fires after 2017.** Annual ignitions fall in every municipality: by about 80% in Guimarães and Braga, 64% in Guarda and 26% in Faro. The change is not limited to one place, which makes it more convincing as a general pattern than as a local fluctuation.
- **Municipalities remain different, but get closer.** Guimarães and Braga keep the highest ignition density and Guarda the lowest, but the gap shrinks as the high-density municipalities fall toward the others.
- **A calmer, more stable period.** Before 2017 the combined annual total ranged from 220 to 1,467 ignitions; after 2017 it stays between 140 and 325.
- **Shorter fires, fewer very large fires per year.** The median duration falls from 1.2 to 0.9 hours, and fires above 100 ha drop from about 4 to about 1 per year. The typical fire stays very small in both periods, so the change shows up mainly in the tail of the distribution.
- **No visible gain in response time.** The median time to first intervention rises slightly (10 → 12 minutes), so the shorter durations cannot be linked to faster first response from these plots alone.

### What changes in the type of fire?

- **More intentional fires among known human causes.** The intentional share rises in all four municipalities, most strongly in Braga (35% → 64%) and Guimarães (19% → 43%).
- **Better recording changes how this looks.** Most fires had no cause attributed before 2017 (55-96% depending on the municipality), and this share drops a lot afterwards. Part of the rise in intentional fires therefore comes from more fires being investigated, not only from a real shift in behaviour. Guarda, the only municipality with good coverage in both periods, shows the smallest change.
- **Municipalities differ in character.** Among ignitions with a known human cause, negligent causes predominate in Faro and Guarda, while Braga moves toward intentional causes.
- **Time of day offers a hint, not a result.** After 2017, intentional fires are more common at night and in the early morning, and negligent fires during the day, which is consistent with, but does not demonstrate, different temporal patterns by cause. The small night-time groups make this suggestive only.

### How to interpret the change?

- **A gradual decline, with 2017 as a descriptive transition point.** Ignitions were already falling before 2017, but the decline was irregular, with sharp peaks and large swings from one year to the next. After 2017 every year stays low and the series is steady, with no return to the earlier peaks. No sudden break appears at the transition year, so this analysis does not establish a structural break or a causal role for the 2017 crisis.
- **Shared yearly patterns suggest external drivers.** Braga and Guimarães rise and fall in the same years, which points to conditions common to the region, such as weather, that act regardless of the period.

### Limitations to keep in mind

- **Missing information is unevenly spread.** Causes are mostly unknown before 2017, and intervention times are missing for all of 2001-2007, so the earlier period is described much less completely than the later one.
- **The periods are unequal.** The pre-2017 period is twice as long and has far more records, so it naturally shows more extreme values. Comparisons of outliers and maximums should be read cautiously.
- **Few observations for annual comparisons.** Annual analyses rely on 16 and 8 years per municipality, and consecutive years are not independent.

### Overall message

Wildfire activity in these four municipalities is lower, more stable, and made up of shorter fires after 2017, with fewer very large fires per year. Among fires with a known human cause, intentional ones become more prominent. The evidence points to a long, gradual decline rather than a sudden shift at 2017, and data gaps, especially in cause and intervention time, limit how far the conclusions can go.

# Draw Inferences

The exploratory analysis described the data; this section tests whether the observed patterns are strong enough to support conclusions about the three research questions. All tests use a significance level of α = 0.05, and the year 2017 is excluded from every before-and-after comparison.

Each research question follows the same structure: research question and variables, hypotheses, data preparation, statistical test, results, assumption checks, visualization, and interpretation of the statistical results. The calculations use the existing `scoped_data` dataset and the municipality-year table already built in the notebook.


```python
# Shared settings and helpers for the inferential analysis.
# SciPy supplies the sampling distributions used by the tests below.
from scipy import stats

ALPHA = 0.05
Z_CRITICAL = stats.norm.ppf(1 - ALPHA / 2)
MIN_EXPECTED_COUNT = 10


def format_p_value(p_value):
    """Format a p-value, showing very small values as '< 0.001'."""
    return "< 0.001" if p_value < 0.001 else f"{p_value:.3f}"

def format_p_text(p_value):
    """Format a p-value for inline text, including the relational symbol."""
    return "p < 0.001" if p_value < 0.001 else f"p = {p_value:.3f}"


def hypothesis_decision(p_value, alpha=ALPHA):
    """Return the decision for a two-sided test at the chosen alpha."""
    return "Reject H₀" if p_value < alpha else "Do not reject H₀"


def wald_proportion_ci(successes, total, z=Z_CRITICAL):
    """95% Wald confidence interval for one proportion."""
    proportion = successes / total
    standard_error = np.sqrt(proportion * (1 - proportion) / total)
    return proportion - z * standard_error, proportion + z * standard_error
```

## RQ1: Proportion of Intentional Ignitions

**Research question.** Did the proportion of intentional ignitions among human-caused ignitions change between 2001-2016 and 2018-2025?

- **Variable:** cause type.
- **Groups:** period (before / after 2017).
- **Test:** two-proportion z-test (compares two percentages).

We define human-caused fires operationally as records classified as intentional or negligent, and use only those records. Natural, reignition and unknown causes are left out, and so is 2017. The percentage is therefore "intentional fires out of intentional + negligent fires", not out of all fires.


### Hypotheses

Let p₁ be the share of intentional fires among human-caused fires in 2001-2016, and p₂ the share in 2018-2025.

- **H₀ (no change):** p₁ = p₂. The share of intentional fires is the same in both periods.
- **H₁ (change):** p₁ ≠ p₂. The share is different.

The significance level is 0.05.


### Data Preparation

We add up the intentional and negligent fires of the four municipalities for each period, so we compare just two percentages: before and after 2017.

A table by municipality is also shown, only to show what the pooled percentages are made of. It is not used in the test.


```python
# RQ1 uses only the two human-caused categories named in the research question.
RQ1_CAUSES = ["INTENCIONAL", "NEGLIGENTE"]

rq1_data = scoped_data.loc[
    scoped_data["CAUSE"].isin(RQ1_CAUSES)
    & scoped_data["PERIOD"].isin(COMPARISON_PERIODS),
    ["MUNICIPALITY", "PERIOD", "CAUSE"],
].copy()
rq1_data["INTENTIONAL"] = rq1_data["CAUSE"].eq("INTENCIONAL").astype(int)

# Validate the subset before calculating anything.
if set(rq1_data["CAUSE"]) != set(RQ1_CAUSES):
    raise ValueError("RQ1 contains causes outside the intentional/negligent subset.")
if (rq1_data["PERIOD"].astype(str) == "2017").any():
    raise ValueError("2017 must be excluded from RQ1.")
if set(rq1_data["PERIOD"].astype(str)) != set(COMPARISON_PERIODS):
    raise ValueError("RQ1 does not contain both comparison periods.")

rq1_counts = (
    rq1_data.groupby("PERIOD", observed=True)["INTENTIONAL"]
    .agg(INTENTIONAL="sum", TOTAL="count")
    .reindex(COMPARISON_PERIODS)
)
rq1_counts["NEGLIGENT"] = rq1_counts["TOTAL"] - rq1_counts["INTENTIONAL"]
rq1_counts = rq1_counts.reset_index()

# Readable count table. The denominator is intentional + negligent.
rq1_counts_table = pd.DataFrame({
    "Period": rq1_counts["PERIOD"].map(PERIOD_LABELS),
    "Intentional": rq1_counts["INTENTIONAL"],
    "Negligent": rq1_counts["NEGLIGENT"],
    "Total human-caused": rq1_counts["TOTAL"],
    "Intentional share (%)": (
        100 * rq1_counts["INTENTIONAL"] / rq1_counts["TOTAL"]
    ).round(2),
})
display(rq1_counts_table)

print(
    f"RQ1 subset: {len(rq1_data):,} records; "
    f"{int(rq1_counts['INTENTIONAL'].sum()):,} intentional and "
    f"{int(rq1_counts['NEGLIGENT'].sum()):,} negligent. "
    "2017 is excluded."
)

# Composition of the pooled comparison by municipality and period.
# This table is descriptive; no municipality-level hypothesis test is performed.
rq1_municipality_counts = (
    rq1_data.groupby(["MUNICIPALITY", "PERIOD"], observed=True)["INTENTIONAL"]
    .agg(INTENTIONAL="sum", TOTAL="count")
    .reindex(
        pd.MultiIndex.from_product(
            [MUNICIPALITIES, COMPARISON_PERIODS],
            names=["MUNICIPALITY", "PERIOD"],
        ),
        fill_value=0,
    )
    .reset_index()
)
rq1_municipality_counts["NEGLIGENT"] = (
    rq1_municipality_counts["TOTAL"]
    - rq1_municipality_counts["INTENTIONAL"]
)

rq1_municipality_table = pd.DataFrame({
    "Municipality": rq1_municipality_counts["MUNICIPALITY"].map(
        MUNICIPALITY_LABELS
    ),
    "Period": rq1_municipality_counts["PERIOD"].map(PERIOD_LABELS),
    "Intentional": rq1_municipality_counts["INTENTIONAL"],
    "Negligent": rq1_municipality_counts["NEGLIGENT"],
    "Total human-caused": rq1_municipality_counts["TOTAL"],
    "Intentional share (%)": (
        100 * rq1_municipality_counts["INTENTIONAL"]
        / rq1_municipality_counts["TOTAL"]
    ).round(2),
}).reset_index(drop=True)
display(rq1_municipality_table)

rq1_pre_total = int(
    rq1_municipality_counts.loc[
        rq1_municipality_counts["PERIOD"] == "2001-2016", "TOTAL"
    ].sum()
)
rq1_pre_guarda = int(
    rq1_municipality_counts.loc[
        (rq1_municipality_counts["PERIOD"] == "2001-2016")
        & (rq1_municipality_counts["MUNICIPALITY"] == "GUARDA"),
        "TOTAL",
    ].iloc[0]
)
rq1_post_total = int(
    rq1_municipality_counts.loc[
        rq1_municipality_counts["PERIOD"] == "2018-2025", "TOTAL"
    ].sum()
)
rq1_post_guimaraes = int(
    rq1_municipality_counts.loc[
        (rq1_municipality_counts["PERIOD"] == "2018-2025")
        & (rq1_municipality_counts["MUNICIPALITY"] == "GUIMARAES"),
        "TOTAL",
    ].iloc[0]
)
print(
    "Composition of the pooled comparison: "
    f"Guarda supplies {rq1_pre_guarda:,} of {rq1_pre_total:,} "
    f"({100 * rq1_pre_guarda / rq1_pre_total:.1f}%) pre-2017 records, "
    f"whereas Guimarães supplies {rq1_post_guimaraes:,} of "
    f"{rq1_post_total:,} ({100 * rq1_post_guimaraes / rq1_post_total:.1f}%) "
    "post-2017 records."
)
```

### Statistical Test

The test asks: *is the gap between the two percentages bigger than what we would expect from chance alone?*

z = (difference between the two shares) ÷ (random variation expected if nothing had changed)

- The larger z is, the less likely the gap is due to chance.
- The **p-value** is the probability of seeing a gap this large if there was really no change.
- If p < 0.05, we reject H₀ (we conclude the share changed).

We also give a **95% confidence interval**: a range of plausible values for the real difference.


### Results

The earlier tables report the pooled counts and the municipality × period composition. The table below reports the difference between the periods, the test statistic, the p-value and the confidence interval.


```python
# Two-proportion z-test with a pooled standard error under H0.
rq1_lookup = rq1_counts.set_index("PERIOD")
rq1_x_pre = int(rq1_lookup.loc["2001-2016", "INTENTIONAL"])
rq1_n_pre = int(rq1_lookup.loc["2001-2016", "TOTAL"])
rq1_x_post = int(rq1_lookup.loc["2018-2025", "INTENTIONAL"])
rq1_n_post = int(rq1_lookup.loc["2018-2025", "TOTAL"])

rq1_p_pre = rq1_x_pre / rq1_n_pre
rq1_p_post = rq1_x_post / rq1_n_post
rq1_diff = rq1_p_post - rq1_p_pre

rq1_pooled_p = (rq1_x_pre + rq1_x_post) / (rq1_n_pre + rq1_n_post)
rq1_pooled_se = np.sqrt(
    rq1_pooled_p * (1 - rq1_pooled_p) * (1 / rq1_n_pre + 1 / rq1_n_post)
)
rq1_z = rq1_diff / rq1_pooled_se
rq1_p_value = 2 * stats.norm.sf(abs(rq1_z))

# The library implementation is used only as a cross-check of the hand calculation.
_, rq1_library_p, _, _ = stats.chi2_contingency(
    np.array([
        [rq1_x_pre, rq1_n_pre - rq1_x_pre],
        [rq1_x_post, rq1_n_post - rq1_x_post],
    ]),
    correction=False,
)
if not np.isclose(rq1_p_value, rq1_library_p):
    raise ValueError("The hand-calculated and library p-values do not agree.")

# Unpooled standard error for the confidence interval of the difference.
rq1_unpooled_se = np.sqrt(
    rq1_p_pre * (1 - rq1_p_pre) / rq1_n_pre
    + rq1_p_post * (1 - rq1_p_post) / rq1_n_post
)
rq1_ci_low = rq1_diff - Z_CRITICAL * rq1_unpooled_se
rq1_ci_high = rq1_diff + Z_CRITICAL * rq1_unpooled_se
rq1_decision = hypothesis_decision(rq1_p_value)

rq1_test_table = pd.DataFrame({
    "Statistic": [
        "Difference in intentional share (post − pre)",
        "Pooled proportion under H₀",
        "Standard error under H₀",
        "z statistic",
        "P-value (two-sided)",
        "95% CI for the difference",
        "Decision at α = 0.05",
    ],
    "Value": [
        f"{100 * rq1_diff:+.2f} p.p.",
        f"{rq1_pooled_p:.4f}",
        f"{rq1_pooled_se:.6f}",
        f"{rq1_z:.3f}",
        format_p_value(rq1_p_value),
        f"[{100 * rq1_ci_low:+.2f}; {100 * rq1_ci_high:+.2f}] p.p.",
        rq1_decision,
    ],
})
display(rq1_test_table)
```

### Assumption Checks

The test works well if:

1. **There are enough fires in each group** (at least 5 expected in each cell of the table). This is clearly met, see the table below.
2. **Fires are independent of each other.** This is only approximately true. Fires in the same place and year share weather and conditions, and the test ignores this, so the result may look a little more certain than it really is.


```python
# Expected counts under H0, using the pooled proportion for both periods.
rq1_expected = pd.DataFrame({
    "Period": [PERIOD_LABELS[p] for p in COMPARISON_PERIODS],
    "Intentional (observed)": [rq1_x_pre, rq1_x_post],
    "Negligent (observed)": [
        rq1_n_pre - rq1_x_pre,
        rq1_n_post - rq1_x_post,
    ],
    "Expected intentional under H₀": [
        rq1_pooled_p * rq1_n_pre,
        rq1_pooled_p * rq1_n_post,
    ],
    "Expected negligent under H₀": [
        (1 - rq1_pooled_p) * rq1_n_pre,
        (1 - rq1_pooled_p) * rq1_n_post,
    ],
})
rq1_expected["Minimum expected count"] = rq1_expected[
    ["Expected intentional under H₀", "Expected negligent under H₀"]
].min(axis=1)
rq1_expected["Condition met (≥ 5)"] = np.where(
    rq1_expected["Minimum expected count"] >= MIN_EXPECTED_COUNT, "Yes", "No"
)
display(rq1_expected.round(2))

if not (rq1_expected["Minimum expected count"] >= MIN_EXPECTED_COUNT).all():
    raise ValueError("The expected-count condition for the z-test is not met.")

print(
    f"Smallest expected count: "
    f"{rq1_expected['Minimum expected count'].min():.2f}. "
    "The expected-count condition is met in both periods."
)
```

## RQ2: Annual number of ignitions

**Research question.** Did the mean annual number of fire ignitions differ between 2001-2016 and 2018-2025?

- **Variable:** number of ignitions per year (the four municipalities added together).
- **Groups:** period (before / after 2017).
- **Test:** Welch's t-test (compares two averages).

All fires are used, whatever their cause. We compare fires **per year** (not totals) because the periods have different lengths (16 vs 8 years).


### Hypotheses

Let $\mu_1$ be the mean annual number of ignitions in 2001-2016 and $\mu_2$ the mean in 2018-2025.

- **H₀:** $\mu_1 = \mu_2$. The mean annual number of ignitions is equal in both periods.
- **H₁:** $\mu_1 \neq \mu_2$. The mean annual number of ignitions differs between periods.

The alternative is two-sided and the significance level is α = 0.05.


### Data preparation

For each year we add up the ignitions of Braga, Guimarães, Guarda and Faro, which gives one number per year: 16 years before and 8 years after (2017 excluded).

We checked the yearly sums against the original data, and no year is missing.


```python
# One observation per year: annual ignitions summed across the four municipalities.
rq2_years = list(range(2001, 2017)) + list(range(2018, 2026))
if len(rq2_years) != 24 or 2017 in rq2_years:
    raise ValueError("The RQ2 year list must contain 24 years and exclude 2017.")
if len(annual_ignition_counts) != len(MUNICIPALITIES) * 25:
    raise ValueError("The municipality-year table does not cover all 25 years.")

rq2_annual = (
    annual_ignition_counts.loc[
        annual_ignition_counts["ANALYSIS_YEAR"].isin(rq2_years)
    ]
    .groupby("ANALYSIS_YEAR", observed=True)["IGNITIONS"]
    .sum()
    .reindex(rq2_years)
    .rename("IGNITIONS")
    .reset_index()
)
if rq2_annual["IGNITIONS"].isna().any():
    raise ValueError("At least one expected year is missing from the annual series.")
rq2_annual["PERIOD"] = np.where(
    rq2_annual["ANALYSIS_YEAR"] <= 2016, "2001-2016", "2018-2025"
)
if set(rq2_annual["PERIOD"]) != set(COMPARISON_PERIODS):
    raise ValueError("The annual series does not contain both comparison periods.")

# Independent spot checks: the year totals must match the canonical dataset.
for check_year in (2001, 2016, 2018, 2025):
    expected_total = int((scoped_data["ANALYSIS_YEAR"] == check_year).sum())
    observed_total = int(
        rq2_annual.loc[
            rq2_annual["ANALYSIS_YEAR"] == check_year, "IGNITIONS"
        ].iloc[0]
    )
    if observed_total != expected_total:
        raise ValueError(
            f"Annual aggregation failed for {check_year}: "
            f"{observed_total} != {expected_total}."
        )

rq2_annual_table = rq2_annual.rename(columns={
    "ANALYSIS_YEAR": "Year", "PERIOD": "Period", "IGNITIONS": "Ignitions",
})
rq2_annual_table["Period"] = rq2_annual_table["Period"].map(PERIOD_LABELS)
display(rq2_annual_table)

rq2_counts_by_period = rq2_annual.groupby("PERIOD", observed=True).size()
print(
    f"Annual observations: {rq2_counts_by_period.get('2001-2016', 0)} "
    f"for 2001-2016 and {rq2_counts_by_period.get('2018-2025', 0)} "
    "for 2018-2025. Aggregation spot checks passed."
)
```

### Statistical Test

The t-test compares two averages and asks whether the gap between them is bigger than chance would explain. We use **Welch's version** because the two periods vary very differently from year to year, and this version still works in that case.

$$
t = \frac{\text{difference between the two averages}}{\text{random variation expected in that difference}}
$$

The p-value is read the same way as in RQ1. We also give a 95% confidence interval for the difference. To describe how big the difference is, we use that interval and the percentage drop, not a standard effect size (the "before" years vary so much more than the "after" years that a single combined measure would be misleading).


### Results

The descriptive table reports the annual observations for each period. The second table reports the Welch test and the confidence interval for the mean difference.


```python
# Descriptive statistics for the annual series in each period.
rq2_pre = rq2_annual.loc[
    rq2_annual["PERIOD"] == "2001-2016", "IGNITIONS"
].to_numpy()
rq2_post = rq2_annual.loc[
    rq2_annual["PERIOD"] == "2018-2025", "IGNITIONS"
].to_numpy()

def describe_annual(values):
    """Return the descriptive statistics reported for RQ2."""
    return {
        "n (years)": len(values),
        "Mean": values.mean(),
        "Standard deviation": values.std(ddof=1),
        "Median": np.median(values),
        "Minimum": values.min(),
        "Maximum": values.max(),
    }

rq2_summaries = {
    period: describe_annual(
        rq2_annual.loc[rq2_annual["PERIOD"] == period, "IGNITIONS"].to_numpy()
    )
    for period in COMPARISON_PERIODS
}

def format_annual_summary(summary):
    """Format one period's descriptive statistics with consistent rounding."""
    return [
        f"{summary['n (years)']}",
        f"{summary['Mean']:.2f}",
        f"{summary['Standard deviation']:.2f}",
        f"{summary['Median']:.2f}",
        f"{summary['Minimum']:.2f}",
        f"{summary['Maximum']:.2f}",
    ]

rq2_descriptives = pd.DataFrame({
    "Statistic": [
        "n (years)",
        "Mean",
        "Standard deviation",
        "Median",
        "Minimum",
        "Maximum",
    ],
    PERIOD_LABELS["2001-2016"]: format_annual_summary(
        rq2_summaries["2001-2016"]
    ),
    PERIOD_LABELS["2018-2025"]: format_annual_summary(
        rq2_summaries["2018-2025"]
    ),
}).set_index("Statistic")
display(rq2_descriptives)

# Welch's independent-samples t-test.
rq2_n_pre = len(rq2_pre)
rq2_n_post = len(rq2_post)
rq2_var_pre = rq2_pre.var(ddof=1)
rq2_var_post = rq2_post.var(ddof=1)
rq2_diff = rq2_pre.mean() - rq2_post.mean()
rq2_se = np.sqrt(rq2_var_pre / rq2_n_pre + rq2_var_post / rq2_n_post)
rq2_df = rq2_se**4 / (
    (rq2_var_pre / rq2_n_pre) ** 2 / (rq2_n_pre - 1)
    + (rq2_var_post / rq2_n_post) ** 2 / (rq2_n_post - 1)
)
rq2_t = rq2_diff / rq2_se
rq2_p_value = 2 * stats.t.sf(abs(rq2_t), rq2_df)
rq2_t_critical = stats.t.ppf(1 - ALPHA / 2, rq2_df)
rq2_ci_low = rq2_diff - rq2_t_critical * rq2_se
rq2_ci_high = rq2_diff + rq2_t_critical * rq2_se
rq2_decision = hypothesis_decision(rq2_p_value)

# Cross-check the hand calculation against SciPy.
rq2_library = stats.ttest_ind(rq2_pre, rq2_post, equal_var=False)
if not (
    np.isclose(rq2_t, rq2_library.statistic)
    and np.isclose(rq2_p_value, rq2_library.pvalue)
    and np.isclose(rq2_df, rq2_library.df)
):
    raise ValueError("The hand-calculated and SciPy Welch tests do not agree.")

rq2_test_table = pd.DataFrame({
    "Statistic": [
        "Mean difference (pre − post)",
        "Standard error of the difference",
        "Welch t statistic",
        "Welch degrees of freedom",
        "P-value (two-sided)",
        "95% CI for the mean difference",
        "Decision at α = 0.05",
    ],
    "Value": [
        f"{rq2_diff:.2f}",
        f"{rq2_se:.2f}",
        f"{rq2_t:.3f}",
        f"{rq2_df:.3f}",
        format_p_value(rq2_p_value),
        f"[{rq2_ci_low:.2f}; {rq2_ci_high:.2f}]",
        rq2_decision,
    ],
})
display(rq2_test_table)
```

### Assumption Checks

The t-test works best if the years are independent and the data in each period have no extreme outliers. Because both periods have fewer than 30 years, we look for clear outliers (the box plot below also shows every year).

- **Normality (outlier check):** the table below counts outliers by the IQR rule for each period. With only 8 years after 2017 this check is limited.
- **Independence:** not fully true. Neighbouring years share weather and fire management, and the municipalities share regional conditions. This can make the result look more certain than it is. The test is an approximate comparison of two averages, not a full time-series analysis.


```python
# Outlier check within each period (rule of thumb for n < 30: look for clear outliers).
rq2_assumption_rows = []
for period, values in [
    ("2001-2016", rq2_pre), ("2018-2025", rq2_post)
]:
    q1, q3 = np.percentile(values, [25, 75])
    iqr = q3 - q1
    outliers = int(
        ((values < q1 - 1.5 * iqr) | (values > q3 + 1.5 * iqr)).sum()
    )
    rq2_assumption_rows.append({
        "Period": PERIOD_LABELS[period],
        "n": len(values),
        "IQR outliers": outliers,
    })
display(pd.DataFrame(rq2_assumption_rows))
```

## RQ3: Differences between municipalities

**Research question.** Did the mean annual number of fire ignitions differ among Braga, Guimarães, Guarda and Faro?

- **Variable:** number of ignitions per year in each municipality.
- **Groups:** municipality.
- **Test:** one-way ANOVA (compares the averages of more than two groups at once).

All fires are used, whatever their cause. The years 2001-2016 and 2018-2025 are included, and 2017 is excluded.


### Hypotheses

$$
H_0: \mu_{\text{Braga}} = \mu_{\text{Guimarães}} = \mu_{\text{Guarda}} = \mu_{\text{Faro}}
$$

$$
H_1: \text{at least one } \mu_i \text{ is different}
$$

The null hypothesis states that the mean annual number of ignitions is equal across the four municipalities. The alternative states that at least one municipality has a different mean annual number of ignitions.

The test is two-sided in the sense that any departure from equality is of interest, and the significance level is α = 0.05.


### Data Preparation

Each observation is one municipality in one year: 24 years per municipality, 96 observations in total. Every municipality-year has records, so there are no missing years to fix.


```python
# Municipality-year observations for the two comparison periods.
rq3_annual = annual_ignition_counts.loc[
    annual_ignition_counts["ANALYSIS_YEAR"].isin(rq2_years),
    ["MUNICIPALITY", "ANALYSIS_YEAR", "IGNITIONS"],
].copy()
rq3_annual["PERIOD"] = np.where(
    rq3_annual["ANALYSIS_YEAR"] <= 2016, "2001-2016", "2018-2025"
)

if (rq3_annual["ANALYSIS_YEAR"] == 2017).any():
    raise ValueError("2017 must be excluded from RQ3.")
if len(rq3_annual) != len(MUNICIPALITIES) * 24:
    raise ValueError("RQ3 does not have 24 observations per municipality.")
if set(rq3_annual["MUNICIPALITY"]) != set(MUNICIPALITIES):
    raise ValueError("RQ3 does not contain all four municipalities.")
if rq3_annual["IGNITIONS"].isna().any():
    raise ValueError("RQ3 contains missing annual ignition counts.")

rq3_observation_counts = (
    rq3_annual.groupby("MUNICIPALITY").size().reindex(MUNICIPALITIES)
)
if not (rq3_observation_counts == 24).all():
    raise ValueError("At least one municipality does not have 24 annual observations.")

rq3_descriptives = (
    rq3_annual.groupby("MUNICIPALITY")["IGNITIONS"]
    .agg(["count", "mean", "std", "median", "min", "max"])
    .reindex(MUNICIPALITIES)
    .rename(columns={
        "count": "n",
        "mean": "Mean",
        "std": "Standard deviation",
        "median": "Median",
        "min": "Minimum",
        "max": "Maximum",
    })
)
rq3_descriptives.index = [
    MUNICIPALITY_LABELS[m] for m in rq3_descriptives.index
]
rq3_descriptives.index.name = "Municipality"
display(rq3_descriptives.round(2))

print(
    f"RQ3 table: {len(rq3_annual)} municipality-year observations "
    f"({len(MUNICIPALITIES)} municipalities × 24 years); 2017 excluded."
)
```

### Statistical Test

ANOVA compares two kinds of variation:

- **Between municipalities:** how different the municipality averages are.
- **Within municipalities:** how much each municipality changes from year to year.

$$
F = \frac{\text{variation between municipalities}}{\text{variation within municipalities}}
$$

If F is large, the municipalities differ more than their normal year-to-year changes would explain, and the p-value is small. We also report **η² (eta-squared)**: the share of all the variation that is explained by the municipality.

ANOVA only says that *at least one* municipality is different, not *which*. To find the pairs, we compare all six pairs with t-tests using the Bonferroni correction (see the post-hoc section below). Welch's ANOVA and the Kruskal-Wallis test are extra checks that do not need equal variation or normal data.


### Results

The ANOVA table partitions the total variation into between-municipality and within-municipality components. The effect-size table reports how much of the variation is associated with municipality, followed by the decision on H₀.


```python
# One-way ANOVA calculated from the sums of squares, then cross-checked with SciPy.
rq3_groups = {
    municipality: rq3_annual.loc[
        rq3_annual["MUNICIPALITY"] == municipality, "IGNITIONS"
    ].to_numpy()
    for municipality in MUNICIPALITIES
}
rq3_all_values = np.concatenate(list(rq3_groups.values()))
rq3_grand_mean = rq3_all_values.mean()

rq3_ss_between = sum(
    len(values) * (values.mean() - rq3_grand_mean) ** 2
    for values in rq3_groups.values()
)
rq3_ss_within = sum(
    ((values - values.mean()) ** 2).sum()
    for values in rq3_groups.values()
)
rq3_ss_total = rq3_ss_between + rq3_ss_within
rq3_df_between = len(rq3_groups) - 1
rq3_df_within = len(rq3_all_values) - len(rq3_groups)
rq3_ms_between = rq3_ss_between / rq3_df_between
rq3_ms_within = rq3_ss_within / rq3_df_within
rq3_f = rq3_ms_between / rq3_ms_within
rq3_p_value = stats.f.sf(rq3_f, rq3_df_between, rq3_df_within)
rq3_eta_squared = rq3_ss_between / rq3_ss_total
rq3_decision = hypothesis_decision(rq3_p_value)

rq3_library = stats.f_oneway(*rq3_groups.values())
if not (
    np.isclose(rq3_f, rq3_library.statistic)
    and np.isclose(rq3_p_value, rq3_library.pvalue)
):
    raise ValueError("The hand-calculated and SciPy ANOVA results do not agree.")
rq3_anova_table = pd.DataFrame({
    "Source": [
        "Between municipalities",
        "Within municipalities (residual)",
        "Total",
    ],
    "Sum of squares": [
        f"{rq3_ss_between:,.2f}",
        f"{rq3_ss_within:,.2f}",
        f"{rq3_ss_total:,.2f}",
    ],
    "df": [rq3_df_between, rq3_df_within, rq3_df_between + rq3_df_within],
    "Mean square": [
        f"{rq3_ms_between:,.2f}",
        f"{rq3_ms_within:,.2f}",
        "",
    ],
    "F": [f"{rq3_f:.3f}", "", ""],
    "P-value": [format_p_value(rq3_p_value), "", ""],
})
display(rq3_anova_table)

rq3_effect_table = pd.DataFrame({
    "Effect size": ["Eta-squared (η²)"],
    "Estimate": [f"{rq3_eta_squared:.3f}"],
    "Interpretation": ["Large"],
})
display(rq3_effect_table)

rq3_decision_table = pd.DataFrame({
    "Decision at α = 0.05": [rq3_decision],
    "Interpretation": [
        "At least one municipality has a different mean annual number of ignitions."
        if rq3_p_value < ALPHA
        else "There is not enough evidence to conclude that the municipality means differ."
    ],
})
display(rq3_decision_table)
```

### Assumption Checks

ANOVA works best if observations are independent, the data in each group are roughly normal, and all groups vary a similar amount.

- **Normality and equal variation are not met** (see the tests below). The year-to-year variation differs a lot between municipalities: the largest is about 276 times the smallest.
- Because every municipality has the same number of years (24), ANOVA handles mild non-normality fairly well. Unequal variation is more serious, so we add Welch's ANOVA and Kruskal-Wallis as extra checks. The standard ANOVA stays as the main test.


```python
# Per-municipality distribution checks.
rq3_assumption_rows = []
for municipality, values in rq3_groups.items():
    shapiro = stats.shapiro(values)
    q1, q3 = np.percentile(values, [25, 75])
    iqr = q3 - q1
    outliers = int(
        ((values < q1 - 1.5 * iqr) | (values > q3 + 1.5 * iqr)).sum()
    )
    rq3_assumption_rows.append({
        "Municipality": MUNICIPALITY_LABELS[municipality],
        "n": len(values),
        "Mean": values.mean(),
        "Standard deviation": values.std(ddof=1),
        "Shapiro-Wilk W": shapiro.statistic,
        "Shapiro-Wilk p": format_p_value(shapiro.pvalue),
        "Skewness": stats.skew(values, bias=False),
        "IQR outliers": outliers,
    })
rq3_distribution_checks = pd.DataFrame(rq3_assumption_rows)
display(rq3_distribution_checks.round(2))

# Levene's test and the pooled-residual normality check.
rq3_levene = stats.levene(*rq3_groups.values())
rq3_residuals = np.concatenate([
    values - values.mean() for values in rq3_groups.values()
])
rq3_residual_shapiro = stats.shapiro(rq3_residuals)
rq3_variance_values = np.array([
    values.var(ddof=1) for values in rq3_groups.values()
])

rq3_assumption_tests = pd.DataFrame({
    "Check": [
        "Levene's test (H₀: equal variances)",
        "Shapiro-Wilk on pooled residuals",
        "Largest / smallest variance",
    ],
    "Result": [
        f"W = {rq3_levene.statistic:.3f}, {format_p_text(rq3_levene.pvalue)}",
        f"W = {rq3_residual_shapiro.statistic:.3f}, {format_p_text(rq3_residual_shapiro.pvalue)}",
        f"{rq3_variance_values.max() / rq3_variance_values.min():.2f}",
    ],
    "Implication": [
        "Equal variances are not supported.",
        "Normality of the residuals is not supported.",
        "The spread differs strongly across municipalities.",
    ],
})
display(rq3_assumption_tests)

# Supplementary robust checks. The one-way ANOVA above remains the primary test.
rq3_ns = np.array([len(values) for values in rq3_groups.values()])
rq3_weights = rq3_ns / rq3_variance_values
rq3_weight_sum = rq3_weights.sum()
rq3_weighted_mean = (
    rq3_weights * np.array([values.mean() for values in rq3_groups.values()])
).sum() / rq3_weight_sum
rq3_welch_numerator = (
    rq3_weights
    * (np.array([values.mean() for values in rq3_groups.values()]) - rq3_weighted_mean) ** 2
).sum() / rq3_df_between
rq3_welch_denominator_term = (
    (1 - rq3_weights / rq3_weight_sum) ** 2 / (rq3_ns - 1)
).sum()
rq3_welch_denominator = (
    1 + 2 * (len(rq3_groups) - 2)
    / (len(rq3_groups) ** 2 - 1)
    * rq3_welch_denominator_term
)
rq3_welch_f = rq3_welch_numerator / rq3_welch_denominator
rq3_welch_df2 = (
    (len(rq3_groups) ** 2 - 1) / (3 * rq3_welch_denominator_term)
)
rq3_welch_p = stats.f.sf(rq3_welch_f, rq3_df_between, rq3_welch_df2)
rq3_kruskal = stats.kruskal(*rq3_groups.values())

rq3_supplementary = pd.DataFrame({
    "Supplementary test": [
        "Welch ANOVA",
        "Kruskal-Wallis",
    ],
    "Statistic": [
        f"F({rq3_df_between}, {rq3_welch_df2:.2f}) = {rq3_welch_f:.3f}",
        f"H({rq3_df_between}) = {rq3_kruskal.statistic:.3f}",
    ],
    "P-value": [
        format_p_value(rq3_welch_p),
        format_p_value(rq3_kruskal.pvalue),
    ],
    "Decision at α = 0.05": [
        hypothesis_decision(rq3_welch_p),
        hypothesis_decision(rq3_kruskal.pvalue),
    ],
})
display(rq3_supplementary)
```

### Post-hoc comparisons (pairwise t-tests with Bonferroni correction)

Since the ANOVA is significant, we compare each pair of municipalities to see which ones differ.

- There are k = 4 municipalities, so K = 4 × 3 / 2 = 6 pairwise comparisons.
- To keep the overall Type 1 error rate under control, each pair is tested at the Bonferroni-corrected significance level α* = 0.05 / 6 ≈ 0.0083.
- Each test uses the pooled standard deviation from the ANOVA table (the square root of the mean square error) and df = n − k = 92.

A pair is different when its two-sided p-value is below α*. Because the municipalities vary by very different amounts (see the assumption checks), results based on the pooled standard deviation should be read with caution.


```python
# Pairwise t-tests with the Bonferroni correction (textbook method).
# Uses the pooled SD from the ANOVA (sqrt of MSE) and df = n - k.
from itertools import combinations

rq3_k_comparisons = len(MUNICIPALITIES) * (len(MUNICIPALITIES) - 1) // 2
rq3_alpha_star = ALPHA / rq3_k_comparisons

rq3_pairwise_rows = []
for municipality_i, municipality_j in combinations(MUNICIPALITIES, 2):
    values_i = rq3_groups[municipality_i]
    values_j = rq3_groups[municipality_j]
    mean_difference = values_i.mean() - values_j.mean()
    pair_se = np.sqrt(
        rq3_ms_within / len(values_i) + rq3_ms_within / len(values_j)
    )
    pair_t = mean_difference / pair_se
    pair_p = 2 * stats.t.sf(abs(pair_t), rq3_df_within)
    rq3_pairwise_rows.append({
        "Comparison": (
            f"{MUNICIPALITY_LABELS[municipality_i]} - "
            f"{MUNICIPALITY_LABELS[municipality_j]}"
        ),
        "Mean difference": mean_difference,
        "SE": pair_se,
        "T": pair_t,
        "df": rq3_df_within,
        "P-value (two-sided)": format_p_value(pair_p),
        f"Significant at α* = {rq3_alpha_star:.4f}": (
            "Yes" if pair_p < rq3_alpha_star else "No"
        ),
    })

rq3_pairwise_table = pd.DataFrame(rq3_pairwise_rows)
display(rq3_pairwise_table.round(3))

rq3_pairwise_significant = int(
    (rq3_pairwise_table.iloc[:, -1] == "Yes").sum()
)
print(
    f"Bonferroni: {rq3_k_comparisons} comparisons, α* = {rq3_alpha_star:.4f}. "
    f"{rq3_pairwise_significant} of {rq3_k_comparisons} pairs are significant."
)
```
